```python
import pandas as pd

# Create a dummy data.csv file to ensure it exists for pd.read_csv
dummy_data = {
    'existing_column': [10, 20, 60, 80, 5, 25],
    'another_column': [10, 50, 70, 90, 100, 120],
    'category_column': ['A', 'B', 'A', 'C', 'B', 'A'],
    'numeric_column': [100, 150, 120, 200, 110, 180]
}
dummy_df = pd.DataFrame(dummy_data)
dummy_df.to_csv('data.csv', index=False)

# Load dataset into a DataFrame
df = pd.read_csv('data.csv')
# Display first and last few rows
print("First 5 rows:\n", df.head())
print("Last 5 rows:\n", df.tail())
# Check data types and general info
df.info()
# Summary statistics
print("Summary statistics:\n", df.describe())
# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)
# Create a new column
df['new_column'] = df['existing_column'] * 2
# Create a Series and perform operations
series = df['existing_column']
print("Series addition:", series + 10)
# Filter rows based on conditions
filtered_df = df[(df['existing_column'] > 50) & (df['another_column'] < 100)]
print("Filtered DataFrame:\n", filtered_df)
# Grouping and aggregation
grouped = df.groupby('category_column')['numeric_column'].mean()
print("Grouped mean:\n", grouped)
# Sorting
df_sorted = df.sort_values(by='numeric_column', ascending=False)
print("Sorted DataFrame:\n", df_sorted)
# Boolean masking
masked_df = df[df['numeric_column'] > df['numeric_column'].median()]

print("Masked DataFrame:\n", masked_df)
# Remove duplicates and drop missing values
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

# Create a new DataFrame with selected columns
subset_df = df[['existing_column', 'another_column', 'numeric_column']]
# Save the new DataFrame to a CSV file
subset_df.to_csv('filtered_data.csv', index=False)
# Compute summary statistics
print("Total sum:", df['numeric_column'].sum())
print("Mean:", df['numeric_column'].mean())
print("Standard Deviation:", df['numeric_column'].std())
```

```python
# ==============================================
# BIVARIATE ANALYSIS: LINEAR & LOGISTIC REGRESSION
# ==============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import r2_score, accuracy_score

# Create a dummy diabetes.csv file to ensure it exists for pd.read_csv
dummy_diabetes_data = {
    'Pregnancies': [6, 1, 8, 1, 0, 5],
    'Glucose': [148, 85, 183, 89, 137, 116],
    'BloodPressure': [72, 66, 64, 66, 40, 74],
    'SkinThickness': [35, 29, 0, 23, 35, 0],
    'Insulin': [0, 0, 0, 94, 168, 0],
    'BMI': [33.6, 26.6, 23.3, 28.1, 43.1, 25.6],
    'DiabetesPedigreeFunction': [0.627, 0.351, 0.672, 0.167, 2.288, 0.201],
    'Age': [50, 31, 32, 21, 33, 30],
    'Outcome': [1, 0, 1, 0, 1, 0]
}
dummy_diabetes_df = pd.DataFrame(dummy_diabetes_data)
dummy_diabetes_df.to_csv('diabetes.csv', index=False)

# ----------------------------
# Load Dataset
# ----------------------------
uci_diabetes = pd.read_csv("diabetes.csv")
pima_diabetes = pd.read_csv("diabetes.csv")

# ----------------------------
# Display Samples
# ----------------------------
print("UCI Diabetes Dataset Sample:\n")
print(uci_diabetes.head())

print("\n\nPima Indians Diabetes Dataset Sample:\n")
print(pima_diabetes.head())


# ===================================================
# Linear Regression Function
# ===================================================
def linear_regression_analysis(df, x_column, y_column):

    X = df[[x_column]]
    Y = df[y_column]

    model = LinearRegression()

    model.fit(X, Y)

    Y_pred = model.predict(X)

    r2 = r2_score(Y, Y_pred)

    print(f"\n\nLinear Regression (Predicting {y_column} using {x_column}):")
    print(f"R² Score: {r2:.4f}")

    plt.figure(figsize=(8,5))

    plt.scatter(
        X,
        Y,
        color='blue',
        label='Actual Data'
    )

    plt.plot(
        X,
        Y_pred,
        color='red',
        linewidth=2,
        label='Regression Line'
    )
    plt.xlabel(x_column)
    plt.ylabel(y_column)

    plt.title(f"Linear Regression: {x_column} vs. {y_column}")

    plt.legend()

    plt.grid(True)

    plt.show()


# ===================================================
# Apply Linear Regression
# ===================================================
linear_regression_analysis(
    uci_diabetes,
    "Glucose",
    "BMI"
)

linear_regression_analysis(
    pima_diabetes,
    "Glucose",
    "BMI"
)


# ===================================================
# Logistic Regression Function
# ===================================================
def logistic_regression_analysis(df):

    features = [
        "Glucose",
        "BloodPressure",
        "BMI",
        "Age"
    ]

    target = "Outcome"
    X = df[features]

    Y = df[target]

    X_train, X_test, Y_train, Y_test = train_test_split(
        X,
        Y,
        test_size=0.20,
        random_state=42
    )

    model = LogisticRegression(max_iter=1000)

    model.fit(X_train, Y_train)

    Y_pred = model.predict(X_test)

    accuracy = accuracy_score(
        Y_test,
        Y_pred
    )

    print("\nLogistic Regression (Predicting Outcome using ['Glucose','BloodPressure','BMI','Age']):")
    print(f"Accuracy Score: {accuracy:.4f}")


# ===================================================
# Apply Logistic Regression
# ===================================================
logistic_regression_analysis(uci_diabetes)
logistic_regression_analysis(pima_diabetes)


# ===================================================
# RESULT
# ===================================================
print("\n\nRESULT:")
print("Linear Regression reveals the relationship between Glucose Level and BMI,")
print("while Logistic Regression predicts Diabetes Presence using selected features.")
print("The obtained R² score indicates the strength of the linear relationship,")
print("and the Accuracy Score reflects the performance of the Logistic Regression model.")
```
