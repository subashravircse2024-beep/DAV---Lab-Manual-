```python
import pandas as pd

# Create a sample DataFrame matching the expected columns
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Age': [24, 22, 25, 23, 26, 21, 24, 22, 25, 23],
    'Department': ['HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR', 'IT', 'Finance', 'HR'],
    'Marks': [60, 85, 75, 45, 92, 55, 70, 88, 62, 78]
}
df_sample = pd.DataFrame(data)

# Save the sample DataFrame to data.csv
df_sample.to_csv('data.csv', index=False)

print("Created 'data.csv' with sample data.")
```

```python
import pandas as pd

df = pd.read_csv("data.csv")

print(df.head())
print(df.tail())

df.info()

print(df.describe())

df.fillna(df.mean(numeric_only=True), inplace=True)

df["Total"] = df["Marks"] * 2

series = df["Marks"]
print(series + 10)

filtered_df = df[(df["Marks"] > 50) & (df["Age"] < 25)]
print(filtered_df)

grouped = df.groupby("Department")["Marks"].mean()
print(grouped)

df_sorted = df.sort_values(by="Marks", ascending=False)
print(df_sorted)

masked_df = df[df["Marks"] > df["Marks"].median()]
print(masked_df)

df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

subset_df = df[["Name", "Marks"]]
subset_df.to_csv("filtered_data.csv", index=False)

print("Total:", df["Marks"].sum())
print("Mean:", df["Marks"].mean())
print("Standard Deviation:", df["Marks"].std())
```
